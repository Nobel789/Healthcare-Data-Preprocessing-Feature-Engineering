⚠️ Detecting Outliers (Important Step)

👉 Outliers = unusual or extreme values (can harm your model)

Detecting Outliers (Important Step)

👉 Outliers = unusual or extreme values (can harm your model)

🧠 Two methods used
1. 📦 IQR Method (most common)
Uses:
Q1 (25%)
Q3 (75%)
Formula:
IQR = Q3 − Q1
Outlier if:
Value < Q1 − 1.5×IQR
Value > Q3 + 1.5×IQR

👉 Best when data is not normally distributed

📊 Z-Score Method
Formula:
Z = (value − mean) / std
Outlier if:
|Z| > 3

👉 Best for normal (bell-shaped) data

```python
def iqr_outliers(series):
    """Return outliers in a pandas Series using the IQR method."""
    q1, q3 = series.quantile([0.25, 0.75])
    iqr = q3 - q1
    return series[(series < q1 - 1.5 * iqr) | (series > q3 + 1.5 * iqr)]

# Function to print outliers for a column
def print_outliers(df, column):
    outliers = iqr_outliers(df[column].dropna())
    print(f"{column} outliers (IQR method):")
    print(f"Found {len(outliers)} outliers")
    if len(outliers) > 0:
        print(f"Outlier values: {outliers.unique()}\n")

# Detect outliers
print_outliers(df, 'Age')
print_outliers(df, 'Glucose_mg_dL')
```

Standardize inconsistent categorical variables

Inconsistent categorical data is common in healthcare due to:
- Multiple data entry personnel with different conventions
- Data merging from different systems
- Typos and abbreviations

You need to standardize:
- **Gender**: Convert M/Male/m → 'Male', F/Female/f → 'Female'
- **Ethnicity**: Standardize capitalization (asian/Asian/ASIAN → 'Asian')
- **Diagnosis_Code**: Fix typos and standardize (OCD/ocd → 'OCD', ANXITY → 'ANX')

This ensures categorical data is **clean, consistent, and machine-readable**.

```python
# Standardize Gender column
print("Before standardization - Gender unique values:")
print(df_clean['Gender'].value_counts(dropna=False))

# Convert to lowercase and strip whitespace for consistent matching
df_clean['Gender'] = df_clean['Gender'].astype(str).str.strip().str.lower()

# Define mapping for known variations
gender_map = {
    'male': 'Male', 'm': 'Male',
    'female': 'Female', 'f': 'Female',
    'other': 'Other',
    'nan': np.nan, 'none': np.nan
}

# Apply mapping
df_clean['Gender'] = df_clean['Gender'].replace({'nan': np.nan})
df_clean['Gender'] = df_clean['Gender'].map(
    lambda x: gender_map.get(x, x.capitalize() if pd.notna(x) else x)
)

print("\nAfter standardization - Gender unique values:")
print(df_clean['Gender'].value_counts(dropna=False))
```

```python
# Standardize Ethnicity column
print("Before standardization - Ethnicity unique values:")
print(df_clean['Ethnicity'].value_counts(dropna=False))

# Clean and capitalize values
df_clean['Ethnicity'] = df_clean['Ethnicity'].astype(str).str.strip().replace({'nan': np.nan})
df_clean['Ethnicity'] = df_clean['Ethnicity'].str.capitalize()

print("\nAfter standardization - Ethnicity unique values:")
print(df_clean['Ethnicity'].value_counts(dropna=False))
```

df_clean['Ethnicity'] selects the column named Ethnicity.
.value_counts(dropna=False) counts how many times each value appears, including missing values (NaN).
This is useful to see what your data looks like before cleaning.


.astype(str) ensures that every value is a string (sometimes numbers or NaNs are stored differently).
.str.strip() removes spaces at the start or end of the string (" African " → "African").
This avoids tiny differences that make the same value appear multiple times.

Sometimes missing values are stored as the string "nan".
.replace({'nan': np.nan}) turns these into actual missing values (np.nan), which pandas understands.


.str.capitalize() makes the first letter uppercase and the rest lowercase.
Example: "asian" → "Asian", "african american" → "African american".

This shows the cleaned column and helps verify the standardization worked.

```python
# Standardize Diagnosis_Code column
print("Before standardization - Diagnosis_Code unique values:")
print(df_clean['Diagnosis_Code'].value_counts(dropna=False))

# Convert to uppercase and fix common typos
df_clean['Diagnosis_Code'] = df_clean['Diagnosis_Code'].astype(str).str.strip().str.upper()
df_clean['Diagnosis_Code'] = df_clean['Diagnosis_Code'].replace({
    'NAN': np.nan,
    'ANXITY': 'ANX',  # Fix typo
    'OCD.': 'OCD'      # Remove trailing period
})

print("\nAfter standardization - Diagnosis_Code unique values:")
print(df_clean['Diagnosis_Code'].value_counts(dropna=False))
```

```python


# Function to convert weight to kilograms
def weight_to_kg(x):
    if pd.isna(x):  # Keep missing values as NaN
        return np.nan

    # If it's already a number, assume kg
    if isinstance(x, (int, float)):
        return float(x)

    # Convert string to lowercase and remove spaces
    s = str(x).strip().lower()

    # Convert lbs to kg
    if 'lb' in s:
        number = float(re.findall(r'\d+\.?\d*', s)[0])
        return round(number * 0.4536, 2)

    # Otherwise, try converting directly to float (assume kg)
    try:
        return float(s)
    except:
        return np.nan

# Apply function to the Weight column
df_clean['Weight_kg'] = df_clean['Weight'].apply(weight_to_kg)

# Show some examples
print("Weight conversion examples:")
display(df_clean[['Weight', 'Weight_kg']].head(10))
```

```python
# Convert height from cm to meters
df_clean['Height_cm'] = pd.to_numeric(df_clean['Height_cm'], errors='coerce')
df_clean['Height_m'] = df_clean['Height_cm'] / 100.0

print("Height converted from cm to meters")
```

```python
# Calculate BMI (Body Mass Index)
def calculate_bmi(weight, height):
    if pd.notna(weight) and pd.notna(height) and height > 0:
        return round(weight / (height ** 2), 2)
    else:
        return np.nan

df_clean['BMI'] = df_clean.apply(lambda row: calculate_bmi(row['Weight_kg'], row['Height_m']), axis=1)

print("BMI calculated successfully")
display(df_clean[['Weight', 'Weight_kg', 'Height_cm', 'Height_m', 'BMI']].head(10))
```

Handling missing values in healthcare data is essential because gaps can come from skipped tests, errors, equipment issues, or privacy restrictions.

Median imputation is preferred over the mean because it:

Is robust to outliers and extreme values.
Represents the middle value, unaffected by extremes.
Preserves the data’s distribution better.
Is simple and fast to compute.

```python
# List of numeric columns to impute
numeric_cols = ['Age', 'Weight_kg', 'Height_cm', 'BMI', 'Glucose_mg_dL']

# Check missing values before imputation
print("Missing values before imputation:")
print(df_clean[numeric_cols].isnull().sum())

# Median imputation
for col in numeric_cols:
    df_clean[col].fillna(df_clean[col].median(), inplace=True)

# Verify missing values are filled
print("\nMissing values after imputation:")
print(df_clean[numeric_cols].isnull().sum())
print("\nAll numeric missing values successfully imputed!")
```

```python
# Parse diagnosis dates with mixed formats
df_clean['Diagnosis_Date_parsed'] = pd.to_datetime(
    df_clean['Diagnosis_Date'],
    errors='coerce',   # Unparseable dates become NaT
    dayfirst=True      # Treat ambiguous dates as day/month/year
)

# Check how many dates were parsed successfully
print("Date parsing results:")
print("Successfully parsed:", df_clean['Diagnosis_Date_parsed'].notna().sum())
print("Failed to parse:", df_clean['Diagnosis_Date_parsed'].isna().sum())

# Display a sample of original vs parsed dates
display(df_clean[['Diagnosis_Date', 'Diagnosis_Date_parsed']].head(10))
```

```python
# Extract year from diagnosis date
df_clean['Diagnosis_Year'] = df_clean['Diagnosis_Date_parsed'].dt.year

# Use the most recent diagnosis date as reference, or today if missing
ref_date = df_clean['Diagnosis_Date_parsed'].max()
if pd.isna(ref_date):
    ref_date = pd.to_datetime("today")

# Calculate days since diagnosis
df_clean['Days_Since_Diagnosis'] = (ref_date - df_clean['Diagnosis_Date_parsed']).dt.days

# Display results
print(f"Reference date for calculating days since diagnosis: {ref_date.date()}")
display(df_clean[['Diagnosis_Date_parsed', 'Diagnosis_Year', 'Days_Since_Diagnosis']].head(10))
```

Encoding categorical column


Encoding categorical variables means converting text categories into numbers so machine learning models can understand them.

One-hot encoding creates a separate 0/1 column for each category.

Example: Diagnosis_Code = ['DEP', 'OCD', 'ANX'] → columns: Diagnosis_Code_DEP, Diagnosis_Code_OCD, Diagnosis_Code_ANX
A patient with 'OCD' would have [0, 1, 0].

Why use one-hot encoding?

Treats all categories equally (no order assumed)
Works with any ML algorithm
Prevents models from misinterpreting category numbers as having value relationships

Note: Label encoding (1, 2, 3…) is only for categories with a natural order.

In practice: Apply one-hot encoding to columns like Gender, Ethnicity, and Diagnosis_Code

```python
# Columns to encode
categorical_cols = ['Diagnosis_Code', 'Gender', 'Ethnicity']

# Check columns before encoding
print("Columns before encoding:", df_clean.columns.tolist())
print("Total columns:", len(df_clean.columns))

# Apply one-hot encoding
df_final = pd.get_dummies(df_clean, columns=categorical_cols, drop_first=False)

# Check columns after encoding
print("\nColumns after encoding:", df_final.columns.tolist())
print("Total columns:", len(df_final.columns))
print("New encoded columns added:", len(df_final.columns) - len(df_clean.columns))
```

Machine learning models cannot use text directly, so you need to convert categories to numbers.
One-hot encoding is the safest choice for non-ordinal categories.
Always verify new columns after encoding to make sure everything looks correct.

Scalling the feature

Scaling numeric features is important because many machine learning algorithms are sensitive to the size of the numbers.

Example: Age ranges 0–100, Glucose ranges 70–500. Without scaling, larger numbers can dominate the model.
Algorithms affected: Logistic Regression, SVM, KNN, Neural Networks, K-Means
Algorithms not affected: Tree-based models like Decision Trees or Random Forests

StandardScaler (Z-score normalization) transforms numbers using:

z=x−meanstd
z=
std
x−mean
	​

Resulting data has mean = 0 and standard deviation = 1.
Best for algorithms that assume normally distributed data.
Values typically range from about -3 to +3.

```python
from sklearn.preprocessing import StandardScaler

# 1. Define numeric columns to scale
numeric_cols = ['Age', 'Weight_kg', 'Height_cm', 'BMI', 'Glucose_mg_dL', 'Days_Since_Diagnosis']
numeric_cols = [col for col in numeric_cols if col in df_final.columns]  # Keep only existing columns

print(f"Scaling {len(numeric_cols)} numeric features:", numeric_cols)

# 2. Fill any remaining missing values with median
df_final[numeric_cols] = df_final[numeric_cols].fillna(df_final[numeric_cols].median())

print("Verified no missing values before scaling:")
print(df_final[numeric_cols].isna().sum())

# 3. Apply StandardScaler
scaler = StandardScaler()
scaled_cols = [col + '_scaled' for col in numeric_cols]
df_final[scaled_cols] = scaler.fit_transform(df_final[numeric_cols])

print("Scaling complete!")

# 4. Check statistics of scaled features
display(df_final[scaled_cols].describe())

# 5. Compare original vs scaled example columns
display(df_final[['Age', 'Age_scaled', 'Glucose_mg_dL', 'Glucose_mg_dL_scaled']].head(10))
```

Always scale numeric features when using algorithms sensitive to feature magnitude (like Logistic Regression, SVM, KNN, Neural Networks).
Keep original values if you want to compare or debug later.
Fill missing values first, because scaling cannot handle NaNs.
